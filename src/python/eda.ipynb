## Unique

## Number

```python
exploded = (
    df['PeopleSoft course ID']
    .dropna()
    .astype(str)
    .str.split(',')
    .explode()
    .str.strip()
)

numeric_cast = pd.to_numeric(exploded, errors='coerce')
non_numeric = exploded[numeric_cast.isna()]
unique_non_numeric = non_numeric.unique()

print(unique_non_numeric)
```

```python
exploded = (
    df['Course number']
    .dropna()
    .astype(str)
    .str.split(',')
    .explode()
    .str.strip()
)

numeric = pd.to_numeric(exploded, errors='coerce')
small = exploded[(numeric.notna()) & (numeric < 1000)]
unique_small = small.unique()

print(unique_small)
```
